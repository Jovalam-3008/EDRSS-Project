```python
import matplotlib.pyplot as plt
import seaborn as sns
from src.database import query_dataframe

daily = query_dataframe('select * from api.page_risk_and_delinquency order by pdate')
daily.head()
```

## 3. Evolución temporal del riesgo

```python
plot_data = daily.dropna(subset=['delinquency_rate']).copy()
plt.figure(figsize=(12, 4))
sns.lineplot(data=plot_data, x='pdate', y='delinquency_rate', hue='dataset_role', marker='o')
plt.title('Evolución temporal de la mora')
plt.ylabel('Tasa de mora')
plt.xlabel('Fecha')
plt.xticks(rotation=45)
plt.tight_layout();
```
